```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
dataset = pd.read_csv("DATASET_LIMPIO_ANEMIA.csv")
```

```python
dataset.info()
```

```python
dataset['quintil_riqueza'].value_counts().plot(kind='bar')
plt.show()
```

```python
dataset['area_residencia'].value_counts().plot(kind='bar')
plt.show()
```

```python
dataset['departamento'].value_counts().head(15).plot(kind='bar')
plt.show()
```

```python
dataset['edad_meses'].value_counts().plot(kind='bar')
plt.show()
```

```python
dataset['sexo_nino'].value_counts().plot(kind='bar')
plt.show()
```

```python
#PREVALENCIA:
```

```python
#Prevalencia de niños con anemia por departamento
#La proporción de niños que tienen anemia respecto al total de niños evaluados en esa región.

df_prevalencia = (dataset.groupby('departamento')['ANEMIA'].mean() * 100).sort_values(ascending=False)

df_prevalencia.plot(kind='bar')
plt.show()
```

```python
#Prevalencia de niños con anemia por edad.

df_prevalencia = (dataset.groupby('area_residencia')['ANEMIA'].mean() * 100).sort_values(ascending=False)

df_prevalencia.plot(kind='bar')
plt.show()
```

```python
#Prevalencia de niños con anemia por edad.

df_prevalencia = (dataset.groupby('edad_meses')['ANEMIA'].mean() * 100).sort_values(ascending=False)

df_prevalencia.plot(kind='bar')
plt.show()
```

```python
#Prevalencia de niños con anemia por area de residencia.

df_prevalencia = (dataset.groupby('quintil_riqueza')['ANEMIA'].mean() * 100).sort_values(ascending=False)

df_prevalencia.plot(kind='bar')
plt.show()
```
